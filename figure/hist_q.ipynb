```python
from matplotlib import pyplot as plt
import numpy as np
import os
import h5py
```

```python
available_chains = os.listdir("catene_emcee/")

print("Available chains:")
for chain in available_chains:
    print(f" - {chain}")
```

```python
# carica i dati da un file .h5
chains = []
for filename in available_chains:
    with h5py.File(f"catene_emcee/{filename}", "r") as f:
        print(f"Loading {filename} with keys: {list(f.keys())}")
        chains.append(f["mcmc"])
        print(f"keys in mcmc: {list(f['mcmc'].keys())}")

t = chains[0]
print(t["accepted"])
```
